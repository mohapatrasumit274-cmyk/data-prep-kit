```python
import re
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
import string

nltk.download('punkt')
nltk.download('punkt_tab')

responses = {
    "greeting": "Hello! I am your AI assistant. How can I help you today?",
    "ai_definition": "Artificial Intelligence is a branch of computer science that focuses on creating systems capable of performing tasks that require human intelligence.",
    "career": "AI offers career roles such as Data Scientist, Machine Learning Engineer, AI Researcher, and Business Analyst.",
    "skills": "To learn AI, you should focus on Python, statistics, machine learning concepts, and problem-solving skills.",
    "bye": "Goodbye! Have a great day.",
    "capabilities": "I am an AI assistant designed to answer questions about AI, careers in AI, and skills needed for AI. I can also help you with basic information related to these topics. I do not have a general knowledge base for factual questions like capital cities.",
    "knowledge_query": "I will look that up in my knowledge base.",
    "knowledge_base_not_found": "I couldn't find an answer to that in my knowledge base.",
    "default": "Sorry, I didn't understand that. Could you please rephrase?"
}

knowledge_base = {
    "capital of france": "Paris",
    "tallest mountain": "Mount Everest",
    "capital of spain": "Madrid",
    "capital of united kingdom": "London"
}

def preprocess_text(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    tokens = word_tokenize(text)
    return tokens

def detect_intent(tokens):
    if any(word in tokens for word in ["hi", "hello", "hey"]):
        return "greeting"
    elif "ai" in tokens or "artificial" in tokens:
        return "ai_definition"
    elif "career" in tokens or "job" in tokens:
        return "career"
    elif "skill" in tokens or "learn" in tokens:
        return "skills"
    # Reordered this condition to be before 'capabilities'
    elif any(word in tokens for word in ["what", "where", "who", "capital", "tallest", "is"]):
        return "knowledge_query"
    elif any(word in tokens for word in ["can", "do", "help", "build", "function", "purpose", "capabilities"]):
        return "capabilities"
    elif "bye" in tokens or "exit" in tokens:
        return "bye"
    else:
        return "default"

def search_knowledge_base(tokens):
    for key, value in knowledge_base.items():
        key_words = key.lower().split()
        # Check if all words in the knowledge base key are present in the user's tokens
        if all(word in tokens for word in key_words):
            return value
    return responses["knowledge_base_not_found"]

def chatbot_response(user_input):
    tokens = preprocess_text(user_input)
    intent = detect_intent(tokens)
    if intent == "knowledge_query":
        return search_knowledge_base(tokens)
    else:
        return responses[intent]

print("AI Assistant: Type 'bye' to exit.")

while True:
    user_input = input("You: ")
    response = chatbot_response(user_input)
    print("AI Assistant:", response)

    if detect_intent(preprocess_text(user_input)) == "bye":
        break
```
